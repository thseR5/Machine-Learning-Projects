```python
# ============================================
# MACHINE LEARNING AND PATTERN RECOGNITION
# LAB WEEK 6 – KNN CLASSIFICATION
# ============================================

# 1) Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# --------------------------------------------
# 2) Load Training and Test Datasets
# --------------------------------------------
train_df = pd.read_csv("tblTrain.csv")
test_df  = pd.read_csv("tblTest.csv")

# --------------------------------------------
# Identify movie-name column safely (if exists)
# --------------------------------------------
name_col = None
for col in ['Movie', 'Title', 'MovieName', 'Name']:
    if col in train_df.columns:
        name_col = col
        break

# --------------------------------------------
# Features and Labels
# --------------------------------------------
X_train = train_df[['Duration', 'Rating']]
y_train = train_df['Genre']

X_test = test_df[['Duration', 'Rating']]
y_test = test_df['Genre']

# --------------------------------------------
# 3) Plot Training Dataset (Annotated)
# --------------------------------------------
plt.figure(figsize=(10, 6))

for genre in ['Action', 'Drama']:
    subset = train_df[train_df['Genre'] == genre]
    plt.scatter(subset['Duration'], subset['Rating'], label=genre)

# Annotation (movie name if exists, else index)
for i in range(len(train_df)):
    label = train_df[name_col][i] if name_col else i
    plt.annotate(label,
                 (train_df['Duration'][i], train_df['Rating'][i]),
                 fontsize=6)

plt.xlabel("Duration (minutes)")
plt.ylabel("IMDB Rating")
plt.title("Indian Movies Dataset – Train Set")
plt.legend()
plt.grid(True)
plt.show()

# --------------------------------------------
# Plot Test Dataset (Annotated)
# --------------------------------------------
plt.figure(figsize=(10, 6))

for genre in ['Action', 'Drama']:
    subset = test_df[test_df['Genre'] == genre]
    plt.scatter(subset['Duration'], subset['Rating'], label=genre)

for i in range(len(test_df)):
    label = test_df[name_col][i] if name_col else i
    plt.annotate(label,
                 (test_df['Duration'][i], test_df['Rating'][i]),
                 fontsize=6)

plt.xlabel("Duration (minutes)")
plt.ylabel("IMDB Rating")
plt.title("Indian Movies Dataset – Test Set")
plt.legend()
plt.grid(True)
plt.show()

# --------------------------------------------
# 4) Model Development and Evaluation
# --------------------------------------------
knn_neighbors = list(range(1, 500, 2))  # [1, 3, 5, ..., 499]
accuracies = []

for k in knn_neighbors:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)

    y_pred = knn.predict(X_test)

    cm = confusion_matrix(y_test, y_pred, labels=['Action', 'Drama'])

    TN, FP, FN, TP = cm.ravel()
    accuracy = (TP + TN) / np.sum(cm)
    accuracies.append(accuracy)

# --------------------------------------------
# 5) Plot Accuracy vs Number of Neighbors (k)
# --------------------------------------------
plt.figure(figsize=(10, 6))
plt.plot(knn_neighbors, accuracies)
plt.xlabel("Number of Neighbors (k)")
plt.ylabel("Accuracy")
plt.title("KNN Accuracy vs Number of Neighbors")
plt.grid(True)
plt.show()

# --------------------------------------------
# 6) Find Best k
# --------------------------------------------
max_accuracy = max(accuracies)
best_k = knn_neighbors[accuracies.index(max_accuracy)]

print("Best k value:", best_k)
print("Maximum Accuracy:", max_accuracy)

# --------------------------------------------
# 7) Confusion Matrix & Performance Metrics
# --------------------------------------------
knn_best = KNeighborsClassifier(n_neighbors=best_k)
knn_best.fit(X_train, y_train)
y_pred_best = knn_best.predict(X_test)

cm_best = confusion_matrix(y_test, y_pred_best, labels=['Action', 'Drama'])

disp = ConfusionMatrixDisplay(confusion_matrix=cm_best,
                              display_labels=['Action', 'Drama'])
disp.plot(cmap=plt.cm.Blues)
plt.title(f"Confusion Matrix – KNN (k = {best_k})")
plt.show()

# --------------------------------------------
# Performance Metrics
# --------------------------------------------
TN, FP, FN, TP = cm_best.ravel()

precision_action = TP / (TP + FP)
recall_action = TP / (TP + FN)

precision_drama = TN / (TN + FN)
recall_drama = TN / (TN + FP)

overall_precision = (precision_action + precision_drama) / 2
overall_recall = (recall_action + recall_drama) / 2
f_score = 2 * overall_precision * overall_recall / (overall_precision + overall_recall)

print("\nPerformance Metrics:")
print("Precision (Action):", precision_action)
print("Recall (Action):", recall_action)
print("Precision (Drama):", precision_drama)
print("Recall (Drama):", recall_drama)
print("Overall Precision:", overall_precision)
print("Overall Recall:", overall_recall)
print("F-score:", f_score)
```

### Model Development and Evaluation:

### Plot Accuracy vs No. of Neighbors (k): 

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# -------- Scale the features --------
sc = StandardScaler()
Xtr = sc.fit_transform(X_train)
Xts = sc.transform(X_test)

# -------- Candidate k values --------
k_values = [k for k in range(1, 500, 2)]
acc_list = []

# -------- Evaluate model for each k --------
for val in k_values:
    model = KNeighborsClassifier(n_neighbors=val)
    model.fit(Xtr, y_train)

    pred = model.predict(Xts)
    mat = confusion_matrix(y_test, pred)

    correct = mat.trace()        # TP + TN
    total = mat.sum()
    acc_list.append(correct / total)

# -------- Plot Accuracy vs k --------
plt.figure(figsize=(10, 5))
plt.plot(k_values, acc_list)
plt.xlabel("Number of Neighbors (k)")
plt.ylabel("Accuracy")
plt.title("KNN Accuracy vs Number of Neighbors")
plt.xlim(0, 500)
plt.ylim(0, 1)
plt.grid(True)
plt.show()

# -------- Determine best k --------
best_accuracy = max(acc_list)
best_position = acc_list.index(best_accuracy)
optimal_k = k_values[best_position]

print(f"Optimal value of k: {optimal_k}")
print(f"Maximum Accuracy: {best_accuracy:.4f}")
```

```python
knn_model = KNeighborsClassifier(n_neighbors=best_k)
knn_model.fit(X_train_scaled, y_train)

# Predictions
y_pred_opt = knn_model.predict(X_test_scaled)

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred_opt, labels=["Action", "Drama"])
tn, fp, fn, tp = cm.ravel()

# Display Confusion Matrix
cm_display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Action", "Drama"]
)
cm_display.plot(cmap="Blues")
plt.title(f"Confusion Matrix - kNN (k = {best_k})")
plt.tight_layout()
plt.show()
```

### Performance Metrics:

```python
tn, fp, fn, tp = cm.ravel()

# Precision and Recall
precision = tp / (tp + fp)
recall = tp / (tp + fn)

# Overall Precision and Overall Recall (same as accuracy)
overall_precision = (tp + tn) / (tp + tn + fp + fn)
overall_recall = overall_precision

# F-score
f_score = (2 * precision * recall) / (precision + recall)

# Print results (4 decimal places)
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"Overall Precision: {overall_precision:.4f}")
print(f"Overall Recall: {overall_recall:.4f}")
print(f"F-score: {f_score:.4f}")
```

## Report:

### Q1: What does the ‘k’ value in KNN determine?
Ans:The value of k determines how many nearest neighbors are considered when making a prediction in the KNN algorithm

### Q2:How is the optimal ‘k’ value in KNN typically chosen?
Ans: The optimal value of k is usually chosen by training the KNN model for different values of k, evaluating each model using a performance metric such as accuracy on a validation or test dataset, and selecting the k that provides the best performance.

### Q3: How does the value of ‘k’ affect variance and bias in KNN?
Ans : A small value of k results in low bias and high variance, making the model sensitive to noise.
A large value of k results in high bias and low variance, leading to smoother decision boundaries and possible underfitting.

### Q4:Is KNN sensitive to outliers, and why?
Ans: Yes, KNN is sensitive to outliers because predictions depend on the closest data points. If an outlier lies close to the data point being classified, it can strongly influence the prediction and lead to incorrect results.

### Q5: How does the scale of features affect the performance of KNN?
Ans:KNN uses distance-based calculations, so features with larger scales can dominate the distance measure. Feature scaling ensures that all features contribute equally, thereby improving the performance of the algorithm.

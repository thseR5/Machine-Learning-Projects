```python
# ============================================
# MACHINE LEARNING AND PATTERN RECOGNITION
# LAB WEEK 6 - KNN CLASSIFICATION
# ============================================

# --------------------------------------------
# 1) Import Required Libraries
# --------------------------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# --------------------------------------------
# 2) Import Training and Test Datasets
# --------------------------------------------
train_data = pd.read_csv("tblTrain.csv")
test_data = pd.read_csv("tblTest.csv")

# Automatically detect movie name column (first column)
movie_column = train_data.columns[0]

# --------------------------------------------
# 3) Plot Training Dataset
# --------------------------------------------
plt.figure(figsize=(10,7))

for genre in train_data['Genre'].unique():
    subset = train_data[train_data['Genre'] == genre]
    plt.scatter(subset['Duration'], subset['Rating'], label=genre)

    # Annotate movie names
    for _, row in subset.iterrows():
        plt.annotate(row[movie_column],
                     (row['Duration'], row['Rating']),
                     fontsize=7)

plt.xlabel("Duration (minutes)")
plt.ylabel("Rating")
plt.title("Training Dataset")
plt.legend()
plt.grid(True)
plt.show()

# --------------------------------------------
# Plot Test Dataset
# --------------------------------------------
plt.figure(figsize=(10,7))

for genre in test_data['Genre'].unique():
    subset = test_data[test_data['Genre'] == genre]
    plt.scatter(subset['Duration'], subset['Rating'], label=genre)

    for _, row in subset.iterrows():
        plt.annotate(row[movie_column],
                     (row['Duration'], row['Rating']),
                     fontsize=7)

plt.xlabel("Duration (minutes)")
plt.ylabel("Rating")
plt.title("Test Dataset")
plt.legend()
plt.grid(True)
plt.show()

# --------------------------------------------
# 4) Model Development and Evaluation
# --------------------------------------------

# Features and Labels
X_train = train_data[['Duration', 'Rating']]
y_train = train_data['Genre']

X_test = test_data[['Duration', 'Rating']]
y_test = test_data['Genre']

# Define k values (odd numbers from 1 to 499)
knn_neighbors = list(range(1, 500, 2))
accuracies = []

for k in knn_neighbors:
    
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    
    y_pred = knn.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)
    
    TN = cm[0][0]
    FP = cm[0][1]
    FN = cm[1][0]
    TP = cm[1][1]
    
    accuracy = (TN + TP) / np.sum(cm)
    accuracies.append(accuracy)

# --------------------------------------------
# 5) Plot Accuracy vs Number of Neighbors (k)
# --------------------------------------------
plt.figure(figsize=(10,6))
plt.plot(knn_neighbors, accuracies)
plt.xlabel("Number of Neighbors (k)")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Number of Neighbors (K)")
plt.grid(True)
plt.show()

# --------------------------------------------
# 6) Find Best k
# --------------------------------------------
max_accuracy = max(accuracies)
best_index = accuracies.index(max_accuracy)
best_k = knn_neighbors[best_index]

print("Best k:", best_k)
print("Maximum Accuracy:", max_accuracy)

# --------------------------------------------
# 7) Final Confusion Matrix and Metrics
# --------------------------------------------
final_knn = KNeighborsClassifier(n_neighbors=best_k)
final_knn.fit(X_train, y_train)
final_pred = final_knn.predict(X_test)

final_cm = confusion_matrix(y_test, final_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=final_cm,
                              display_labels=final_knn.classes_)
disp.plot()
plt.title(f"Confusion Matrix (k = {best_k})")
plt.show()

TN = final_cm[0][0]
FP = final_cm[0][1]
FN = final_cm[1][0]
TP = final_cm[1][1]

# Performance Metrics
precision = TP / (TP + FP)
recall = TP / (TP + FN)
overall_precision = (TP + TN) / (TP + TN + FP + FN)
overall_recall = overall_precision  # same as accuracy
f_score = 2 * (precision * recall) / (precision + recall)

print("Precision:", precision)
print("Recall:", recall)
print("Overall Precision (Accuracy):", overall_precision)
print("Overall Recall:", overall_recall)
print("F-Score:", f_score)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, precision_recall_fscore_support

# 2. Import the training and test datasets
df_train = pd.read_csv('tblTrain.csv')
df_test = pd.read_csv('tblTest.csv')

# Clean column names just in case there are leading/trailing spaces
df_train.columns = df_train.columns.str.strip()
df_test.columns = df_test.columns.str.strip()

# Prepare X and y
X_train = df_train[['Duration', 'Rating']]
y_train = df_train['Genre']

X_test = df_test[['Duration', 'Rating']]
y_test = df_test['Genre']

# 3. Plot the training and test data sets
def plot_dataset(df, title):
    plt.figure(figsize=(12, 8))
    
    # Filter classes
    action_df = df[df['Genre'] == 'Action']
    drama_df = df[df['Genre'] == 'Drama']
    
    # Scatter plots
    plt.scatter(action_df['Duration'], action_df['Rating'], color='red', label='Action', alpha=0.6)
    plt.scatter(drama_df['Duration'], drama_df['Rating'], color='blue', label='Drama', alpha=0.6)
    
    # Annotate points
    for i in range(len(df)):
        plt.annotate(df['Name'].iloc[i], 
                     (df['Duration'].iloc[i], df['Rating'].iloc[i]), 
                     fontsize=6, alpha=0.7)
        
    plt.title(title)
    plt.xlabel('Duration (minutes)')
    plt.ylabel('IMDb Rating')
    plt.legend()
    plt.grid(True)
    plt.show()

# Generate the requested plots
plot_dataset(df_train, 'Indian Movies Dataset - Train Set')
plot_dataset(df_test, 'Indian Movies Dataset - Test Set')

# 4. Model Development and Evaluation
knn_neighbors = list(range(1, 500, 2)) # Generates [1, 3, 5, ..., 499]
accuracies = []

for k in knn_neighbors:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    accuracies.append(accuracy_score(y_test, y_pred))

# 5. Accuracy vs. No. of Neighbors Plot
plt.figure(figsize=(10, 6))
plt.plot(knn_neighbors, accuracies, marker='o', linestyle='-', color='b')
plt.title('Accuracy vs. Number of Neighbors (K)')
plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Accuracy')
plt.grid(True)
plt.show()

# 6. Find the value of k for maximum accuracy
max_accuracy = max(accuracies)
optimal_k_index = accuracies.index(max_accuracy)
optimal_k = knn_neighbors[optimal_k_index]

print(f"Maximum Accuracy: {max_accuracy:.4f}")
print(f"Optimal 'k': {optimal_k}")

# 7. Performance metrics for optimal 'k'
final_knn = KNeighborsClassifier(n_neighbors=optimal_k)
final_knn.fit(X_train, y_train)
y_pred_final = final_knn.predict(X_test)

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred_final, labels=final_knn.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=final_knn.classes_)
disp.plot(cmap=plt.cm.Blues)
plt.title(f'Confusion Matrix - KNN Classification (K: {optimal_k})')
plt.show()

# Calculate Metrics
precision, recall, f_score, _ = precision_recall_fscore_support(y_test, y_pred_final, labels=final_knn.classes_)

print("\n--- Performance Metrics ---")
for i, class_name in enumerate(final_knn.classes_):
    print(f"Class '{class_name}':")
    print(f"  Precision: {precision[i]:.4f}")
    print(f"  Recall:    {recall[i]:.4f}")
    print(f"  F1-Score:  {f_score[i]:.4f}")

# Overall (macro average)
overall_precision = precision.mean()
overall_recall = recall.mean()
overall_f_score = f_score.mean()

print(f"\nOverall Precision: {overall_precision:.4f}")
print(f"Overall Recall:    {overall_recall:.4f}")
print(f"Overall F-Score:   {overall_f_score:.4f}")
```

```python
# Lab Week 6 - Machine Learning and Pattern Recognition
# K-Nearest Neighbors Classification on Indian Movies Dataset

# -------------------------------------------------------
# Step 1: Import Libraries
# -------------------------------------------------------
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# -------------------------------------------------------
# Step 2: Import Training and Test Datasets
# -------------------------------------------------------
train = pd.read_csv('tblTrain.csv')
test  = pd.read_csv('tblTest.csv')

print("Train shape:", train.shape)
print("Test shape:", test.shape)

# -------------------------------------------------------
# Step 3: Plot Training and Test Datasets
# -------------------------------------------------------
def plot_dataset(df, title):
    fig, ax = plt.subplots(figsize=(18, 8))
    
    colors = {'Action': 'blue', 'Drama': 'orange'}
    
    for genre, grp in df.groupby('Genre'):
        ax.scatter(grp['Duration'], grp['Rating'],
                   label=genre, color=colors[genre],
                   marker='*', s=60, alpha=0.7)
    
    # Annotate each data point with movie name
    for _, row in df.iterrows():
        ax.annotate(row['Name'], (row['Duration'], row['Rating']),
                    fontsize=4, alpha=0.7)
    
    ax.set_xlabel('Duration (minutes)')
    ax.set_ylabel('IMDb Rating')
    ax.set_title(title)
    ax.legend()
    plt.tight_layout()
    plt.savefig(f"{title.replace(' ', '_')}.png", dpi=150)
    plt.show()

plot_dataset(train, 'Indian Movies Dataset - Train Set')
plot_dataset(test,  'Indian Movies Dataset - Test Set')

# -------------------------------------------------------
# Step 4: Model Development and Evaluation
# -------------------------------------------------------

# Prepare features and labels
X_train = train[['Duration', 'Rating']].values
y_train = train['Genre'].values

X_test = test[['Duration', 'Rating']].values
y_test = test['Genre'].values

# Define odd k values from 1 to 499
knn_neighbors = list(range(1, 500, 2))

accuracies = []

for k in knn_neighbors:
    # 1. Apply k-NN on training set
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    
    # 2. Get predictions on test set
    y_pred = knn.predict(X_test)
    
    # 3. Compute confusion matrix
    cm = confusion_matrix(y_test, y_pred, labels=['Action', 'Drama'])
    
    # 4. Calculate accuracy: (TN + TP) / Total
    TN = cm[0, 0]  # Action correctly predicted as Action
    TP = cm[1, 1]  # Drama correctly predicted as Drama
    total = cm.sum()
    accuracy = (TN + TP) / total
    accuracies.append(accuracy)

# -------------------------------------------------------
# Step 5: Plot Accuracy vs Number of Neighbors
# -------------------------------------------------------
plt.figure(figsize=(10, 5))
plt.plot(knn_neighbors, accuracies, color='blue', linewidth=1)
plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Accuracy')
plt.title('K-NN Number of Neighbors vs. ML Model Accuracy')
plt.tight_layout()
plt.savefig('Accuracy_vs_K.png', dpi=150)
plt.show()

# -------------------------------------------------------
# Step 6: Find optimal k with maximum accuracy
# -------------------------------------------------------
max_accuracy = max(accuracies)
best_index   = accuracies.index(max_accuracy)
best_k       = knn_neighbors[best_index]

print(f"\nBest k = {best_k}")
print(f"Maximum Accuracy = {max_accuracy:.4f}")

# -------------------------------------------------------
# Step 7: Confusion Matrix and Performance Metrics for best k
# -------------------------------------------------------
knn_best = KNeighborsClassifier(n_neighbors=best_k)
knn_best.fit(X_train, y_train)
y_pred_best = knn_best.predict(X_test)

cm_best = confusion_matrix(y_test, y_pred_best, labels=['Action', 'Drama'])
print("\nConfusion Matrix:\n", cm_best)

# Display confusion matrix
disp = ConfusionMatrixDisplay(confusion_matrix=cm_best,
                              display_labels=['Action', 'Drama'])
fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(ax=ax, colorbar=False, cmap='Blues')
ax.set_xlabel('Predicted Class')
ax.set_ylabel('True Class')
ax.set_title(f'Confusion Matrix - kNN Classification (Number of Neighbors: {best_k})')
plt.tight_layout()
plt.savefig('Confusion_Matrix_Best_K.png', dpi=150)
plt.show()

# Performance Metrics
# cm layout (labels=['Action','Drama']):
#           Pred Action   Pred Drama
# Act True    TN (0,0)    FP (0,1)
# Dra True    FN (1,0)    TP (1,1)

TN = cm_best[0, 0]  # Action correctly classified
FP = cm_best[0, 1]  # Action classified as Drama
FN = cm_best[1, 0]  # Drama classified as Action
TP = cm_best[1, 1]  # Drama correctly classified

# Per-class metrics
precision_action = TN / (TN + FN)          # of predicted Action, how many are Action
precision_drama  = TP / (TP + FP)          # of predicted Drama, how many are Drama

recall_action    = TN / (TN + FP)          # of actual Action, how many predicted Action
recall_drama     = TP / (TP + FN)          # of actual Drama, how many predicted Drama

# Overall metrics
overall_precision = (precision_action + precision_drama) / 2
overall_recall    = (recall_action    + recall_drama)    / 2

F_score = (2 * overall_precision * overall_recall) / (overall_precision + overall_recall)

print("\n--- Performance Metrics ---")
print(f"Precision (Action) : {precision_action:.4f}")
print(f"Precision (Drama)  : {precision_drama:.4f}")
print(f"Recall    (Action) : {recall_action:.4f}")
print(f"Recall    (Drama)  : {recall_drama:.4f}")
print(f"Overall Precision  : {overall_precision:.4f}")
print(f"Overall Recall     : {overall_recall:.4f}")
print(f"F-Score            : {F_score:.4f}")
```

```python
# Lab Week 6 - Machine Learning and Pattern Recognition
# K-Nearest Neighbors Classification on Indian Movies Dataset

# -------------------------------------------------------
# Step 1: Import Libraries
# -------------------------------------------------------
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# -------------------------------------------------------
# Step 2: Import Training and Test Datasets
# -------------------------------------------------------
train = pd.read_csv('tblTrain.csv')
test  = pd.read_csv('tblTest.csv')

print("Train shape:", train.shape)
print("Test shape:", test.shape)

# -------------------------------------------------------
# Step 3: Plot Training and Test Datasets
# -------------------------------------------------------
def plot_dataset(df, title):
    fig, ax = plt.subplots(figsize=(18, 8))

    colors = {'Action': 'blue', 'Drama': 'orange'}

    for genre, grp in df.groupby('Genre'):
        ax.scatter(grp['Duration'], grp['Rating'],
                   label=genre, color=colors[genre],
                   marker='*', s=60, alpha=0.7)

    # Annotate each data point with movie name
    for _, row in df.iterrows():
        ax.annotate(row['Name'], (row['Duration'], row['Rating']),
                    fontsize=4, alpha=0.7)

    ax.set_xlabel('Duration (minutes)')
    ax.set_ylabel('IMDb Rating')
    ax.set_title(title)
    ax.legend()
    plt.tight_layout()
    plt.savefig(f"{title.replace(' ', '_')}.png", dpi=150)
    plt.show()

plot_dataset(train, 'Indian Movies Dataset - Train Set')
plot_dataset(test,  'Indian Movies Dataset - Test Set')

# -------------------------------------------------------
# Step 4: Model Development and Evaluation
# -------------------------------------------------------

# Prepare features and labels
X_train = train[['Duration', 'Rating']].values
y_train = train['Genre'].values

X_test = test[['Duration', 'Rating']].values
y_test = test['Genre'].values

# Define odd k values from 1 to 499
knn_neighbors = list(range(1, 500, 2))

accuracies = []

for k in knn_neighbors:
    # 1. Apply k-NN algorithm on the train set
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)

    # 2. Use the test dataset to find the predictions set
    y_pred = knn.predict(X_test)

    # 3. Get the confusion matrix for each k
    cm = confusion_matrix(y_test, y_pred, labels=['Action', 'Drama'])

    # 4. Calculate accuracy: (TN + TP) / Total
    TN = cm[0, 0]   # Action correctly predicted as Action
    TP = cm[1, 1]   # Drama correctly predicted as Drama
    total = cm.sum()
    accuracy = (TN + TP) / total
    accuracies.append(accuracy)

# -------------------------------------------------------
# Step 5: Plot Accuracy vs Number of Neighbors
# -------------------------------------------------------
plt.figure(figsize=(10, 5))
plt.plot(knn_neighbors, accuracies, color='blue', linewidth=1)
plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Accuracy')
plt.title('K-NN Number of Neighbors vs. ML Model Accuracy')
plt.tight_layout()
plt.savefig('Accuracy_vs_K.png', dpi=150)
plt.show()

# -------------------------------------------------------
# Step 6: Find optimal k with maximum accuracy
# -------------------------------------------------------
# max() finds the highest accuracy
# list.index() returns the FIRST occurrence = smallest k with max accuracy
max_accuracy = max(accuracies)
best_index   = accuracies.index(max_accuracy)
best_k       = knn_neighbors[best_index]

print(f"\nBest k = {best_k}")
print(f"Maximum Accuracy = {max_accuracy:.4f}")

# -------------------------------------------------------
# Step 7: Confusion Matrix and Performance Metrics for best k
# -------------------------------------------------------
knn_best = KNeighborsClassifier(n_neighbors=best_k)
knn_best.fit(X_train, y_train)
y_pred_best = knn_best.predict(X_test)

cm_best = confusion_matrix(y_test, y_pred_best, labels=['Action', 'Drama'])
print("\nConfusion Matrix:\n", cm_best)

# Display confusion matrix
disp = ConfusionMatrixDisplay(confusion_matrix=cm_best,
                              display_labels=['Action', 'Drama'])
fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(ax=ax, colorbar=False, cmap='Blues')
ax.set_xlabel('Predicted Class')
ax.set_ylabel('True Class')
ax.set_title(f'Confusion Matrix - kNN Classification (Number of Neighbors: {best_k})')
plt.tight_layout()
plt.savefig('Confusion_Matrix_Best_K.png', dpi=150)
plt.show()

# -------------------------------------------------------
# Performance Metrics
# Confusion matrix layout with labels=['Action', 'Drama']:
#
#              Pred Action   Pred Drama
# True Action    TN (0,0)    FP (0,1)
# True Drama     FN (1,0)    TP (1,1)
# -------------------------------------------------------

TN = cm_best[0, 0]   # Action correctly classified
FP = cm_best[0, 1]   # Action misclassified as Drama
FN = cm_best[1, 0]   # Drama misclassified as Action
TP = cm_best[1, 1]   # Drama correctly classified

# Per-class Precision: of all predicted in that class, how many are correct
precision_action = TN / (TN + FN)
precision_drama  = TP / (TP + FP)

# Per-class Recall: of all actual in that class, how many were predicted correctly
recall_action    = TN / (TN + FP)
recall_drama     = TP / (TP + FN)

# Overall metrics (macro average)
overall_precision = (precision_action + precision_drama) / 2
overall_recall    = (recall_action    + recall_drama)    / 2

# F-Score (harmonic mean of overall precision and recall)
F_score = (2 * overall_precision * overall_recall) / (overall_precision + overall_recall)

print("\n--- Performance Metrics ---")
print(f"Precision (Action) : {precision_action:.4f}")
print(f"Precision (Drama)  : {precision_drama:.4f}")
print(f"Recall    (Action) : {recall_action:.4f}")
print(f"Recall    (Drama)  : {recall_drama:.4f}")
print(f"Overall Precision  : {overall_precision:.4f}")
print(f"Overall Recall     : {overall_recall:.4f}")
print(f"F-Score            : {F_score:.4f}")
```
